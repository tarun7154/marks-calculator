```python
import marks 
import grades 
import report
```

```python
records= marks.load_records()
print(len(records),"student loaded")
```

```python
def menu():
    print("\n1. Add student")
    print("2. View students")
    print("3. Show report card")
    print("4. Show topper")
    print("5. Show class average")
    print("6. Save")
    print("7. Exit")


def run():
    while True:
        menu()
        choice = input("Enter choice: ").strip()
        if choice == "1":
            marks.add_student(records)
        elif choice == "2":
            marks.view_students(records)
        elif choice == "3":
            report.show_report(records)
        elif choice == "4":
            report.find_topper(records)
        elif choice == "5":
            report.class_average(records)
        elif choice == "6":
            marks.save_records(records)
        elif choice == "7":
            marks.save_records(records)
            print("Bye!")
            break
        else:
            print("Wrong choice, try again.")
```

```python
run()
```
